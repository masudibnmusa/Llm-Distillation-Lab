```python
# Cell 1
import pandas as pd
from distill.evaluation.error_analysis import analyze
from distill.evaluation.teacher_agreement import agreement
RUN = "qwen-1.5b"
res = analyze(RUN)
res

# Cell 2: errors grouped by cause
errs = pd.read_csv(f"../data/runs/{RUN}/errors.csv")
errs.groupby("cause").size().sort_values(ascending=False)

# Cell 3: read the errors in each cause group
for cause, g in errs.groupby("cause"):
    print("\n==", cause, len(g))
    print(g[["text", "gold", "pred", "conf"]].head(8).to_string(index=False))

# Cell 4: is the student copying teacher mistakes?
agreement(RUN)

# Cell 5: calibration
from distill.evaluation.calibration import from_run
from distill.reporting.plots import reliability_diagram
c = from_run(RUN); print("ECE", c["ece"]); reliability_diagram(c["bins"])
```
