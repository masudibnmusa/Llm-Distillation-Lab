```python
# Cell 1
import pandas as pd
from distill.config import TEACHER_LABELS_FILE, read_jsonl
df = pd.DataFrame(read_jsonl(TEACHER_LABELS_FILE))
df["n_words"] = df["text"].str.split().str.len()
df.head()

# Cell 2: agreement distribution and invalid rate
print(df["agreement"].describe())
print("no valid label:", df["label"].isna().mean())

# Cell 3: class balance
df["label"].value_counts().plot.barh(figsize=(6, 10))

# Cell 4: look at the disputed examples by hand
df[df["agreement"] < 0.67].sample(min(20, (df["agreement"] < 0.67).sum()))[["text", "label", "agreement"]]

# Cell 5: spend so far
from distill.teacher.cost_ledger import summary
summary()
```
